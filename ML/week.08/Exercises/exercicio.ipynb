## Exercício

Prepare os seus modelos e submeta as suas previsões para o concurso [Bike Sharing Demand](https://www.kaggle.com/c/bike-sharing-demand/).

Embora este seja um problema de previsão de séries temporais, podemos abordá-lo, para já, com algoritmos clássicos de aprendizagem automática, usando características como a data, a hora e as condições meteorológicas. Esta é uma abordagem inicial válida... mas não devemos usar valores futuros para prever valores passados.
